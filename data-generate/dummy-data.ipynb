```python
from pyspark.sql.functions import expr, current_timestamp

df = (
    spark.range(0,50000)
    .withColumn("event_id", expr("uuid()"))
    .withColumn("user_id", expr("concat('user_',cast(rand() * 1000 as int))"))
    .withColumn("event_type", expr("""
                                    case 
                                     when rand() <0.3 THEN 'page_view'
                                     when rand() <0.6 THEN 'login'
                                     when rand() < 0.8 THEN 'search'
                                     else 'transaction'
                                    end
                                    """))
    .withColumn("event_time", current_timestamp())
    .withColumn("ingestion_time", current_timestamp())
    
)
```

```python
df.display()
```

```python
df.write.mode("append").parquet("/Volumes/maps-data/bronze/raw_ingestion/event_data")
```
